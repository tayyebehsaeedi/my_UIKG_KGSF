# Save all outputs

```python
import zipfile
import glob
import os
from datetime import datetime

# Find all zip files ending with 'out.zip'
zip_files = glob.glob("*out.zip")

# Create a timestamp string
timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")

# Final zip filename with timestamp
combined_zip_name = f"combined_out_zips_{timestamp}.zip"

# Bundle all zips into one container
with zipfile.ZipFile(combined_zip_name, "w") as combined:
    for zf in zip_files:
        combined.write(zf, arcname=os.path.basename(zf))

print(f"Created {combined_zip_name} with {len(zip_files)} zip files.")
```
